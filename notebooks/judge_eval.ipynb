```
import sys, json
sys.path.insert(0, '/content/newsroom-agents/src')
from newsroom import HFLocalLLM
from newsroom.label_eval import evaluate_judge, print_report

llm = HFLocalLLM('Qwen/Qwen2.5-7B-Instruct')
```

```
res = evaluate_judge(llm, 'results/benchmark_diag.json', 'results/labeling/labels_filled.csv')
print_report(res)
json.dump(res, open('results/judge_eval.json', 'w'), ensure_ascii=False, indent=1)
```
